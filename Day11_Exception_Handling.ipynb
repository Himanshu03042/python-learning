```python
# try and except and else error handeling
try:
  number=int(input("enter a number:-"))
  print(number)
except:
  print("invalid error!")
else:
  print("valid number:-",number)


#try and except for tow numbers
try:
  number1=int(input("enter a number1:-"))
  number2=int(input("enter a number2:-"))

except:
  print("invalid error!")
else:
  print("valid numbers sum:-",number1+number2)


#try except&finally
try:
  number3=int(input("enter number3:-"))
  number4=int(input("enter number4:-"))
except ValueError:
  print("invalid numeber:-")
else:
  print(number3+number4)
finally:
  print("program finished")

#finally try except else and finally
try:
  number3=int(input("enter number3:-"))
  number4=int(input("enter number4:-"))
except ValueError:
  print("invalid numeber:-")
else:
  print("valid input")
finally:
  print("THANK YOU")

```

```python
# multiple exception
try:
  num1=int(input("enter the number1:-"))
  num2=int(input("enter the number2:-"))
  result=num1/num2
except ValueError:
  print("invalid number")
except ZeroDivisionError:
  print("can not be divided by zero")
else:
  print("the result is:-",result)

finally:
  print("the program is finished")
```

```python
#explain code
# DAY 11 MINI PROJECT - REAL WORLD STUDENT MARKS CALCULATOR

print("========== STUDENT MARKS CALCULATOR ==========")

try:
    # 1. Student name
    name = input("Enter student name: ")

    # 2. Marks input
    mark1 = int(input("Enter Python marks: "))
    mark2 = int(input("Enter Maths marks: "))
    mark3 = int(input("Enter Physics marks: "))
    mark4 = int(input("Enter Chemistry marks: "))
    mark5 = int(input("Enter English marks: "))

    # 3. Marks validation
    if mark1 < 0 or mark1 > 100:
        raise ValueError("Invalid marks")

    if mark2 < 0 or mark2 > 100:
        raise ValueError("Invalid marks")

    if mark3 < 0 or mark3 > 100:
        raise ValueError("Invalid marks")

    if mark4 < 0 or mark4 > 100:
        raise ValueError("Invalid marks")

    if mark5 < 0 or mark5 > 100:
        raise ValueError("Invalid marks")

    # 4. Total marks
    total = sum([mark1, mark2, mark3, mark4, mark5])

    # 5. Percentage
    percentage = total / 5

    # 6. Individual PASS / FAIL
    print("\n---------- SUBJECT RESULT ----------")

    if mark1 < 40:
        print("Python: FAIL")
    else:
        print("Python: PASS")

    if mark2 < 40:
        print("Maths: FAIL")
    else:
        print("Maths: PASS")

    if mark3 < 40:
        print("Physics: FAIL")
    else:
        print("Physics: PASS")

    if mark4 < 40:
        print("Chemistry: FAIL")
    else:
        print("Chemistry: PASS")

    if mark5 < 40:
        print("English: FAIL")
    else:
        print("English: PASS")

    # 7. Overall PASS / FAIL
    if mark1 < 40 or mark2 < 40 or mark3 < 40 or mark4 < 40 or mark5 < 40:
        result = "FAIL"
    else:
        result = "PASS"

    # 8. Grade
    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    # 9. Highest and lowest marks
    highest = max([mark1, mark2, mark3, mark4, mark5])
    lowest = min([mark1, mark2, mark3, mark4, mark5])

    # 10. Final result
    print("\n========== FINAL RESULT ==========")
    print("Student Name:", name)
    print("Total Marks:", total, "/ 500")
    print("Percentage:", percentage, "%")
    print("Result:", result)
    print("Grade:", grade)
    print("Highest Marks:", highest)
    print("Lowest Marks:", lowest)
    print("==================================")

except ValueError:
    print("Please enter valid marks!")

finally:
    print("The program is finished.")
```
