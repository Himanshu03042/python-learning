```python
try:
  number=int(input("enter the number:-"))
  print(number)
except ValueError:
  print("please enter the valid number")
```

```python
try:
  number=10
  result=number/0
  print(result)
except ZeroDivisionError:
  print("please enter the valid number")



```

```python
try:
  number=int(input("enter the number:-"))
  result=100/number
  print(result)
except ValueError:
  print("please enter the valid number")
except ZeroDivisionError:
  print("can not be divided by zero")
else:
  print("division successful ")
finally:
  print("program execution completed")
```

```python
while True:
  try:
    age=int(input("enter the valid age"))
    if age<=0:
      print("enter the positive age")
      continue
    print("valid age:--",age)
    break
  except ValueError:
    print("please enter the valid number")
```

```python
try:
  num1=float(input("enter the first number:-"))
  num2=float(input("enter the seecond number"))
  print(num1,num2)



  operator=input("enter the operator(+,-,*,/)")
  print(operator)

  if operator=="+":
    print(num1+num2)
  elif operator=="-":
    print(num1-num2)
  elif operator=="*":
    print(num1*num2)
  elif operator=="/":
    print(num1/num2)
  else:
    print("invalid number:")
except ValueError:
    print("please enter the valid numbers")
except ZeroDivisionError:
  print("can not be divided by zero")
```
