```python

#task 5
import random
import math

participants=["himanshu","montu","rahul","naina","nick","vjay","rana","ashu"]
prizes=["RS500 Voucher","headphones","smartwatch","gift hamper"]
winner=random.sample(participants,3)
print("three random winners",winner)
prizes=random.sample(prizes,3)
print("three random prizes",prizes)



winners=random.sample(participants,3)
random_prizes=random.sample(prizes,3)
for winner,prize in zip(winners,random_prizes):
  print("winners:-",winner,"->","prizes:-",prizes)
```

```python
#mini project 2
import random
import math
participants=["himanshu","montu","rahul","naina","nick","vjay","rana","ashu"]
prizes=["RS500 Voucher","headphones","smartwatch","gift hamper"]
random.shuffle(participants)
winner =random.choice(participants)
remaining_participants=participants.copy()
remaining_participants.remove(winner)
backup_winners=random.sample(participants,3)
random.shuffle(prizes)
winner_prize=random.choice(prizes)
remaining_prizes=prizes.copy()
remaining_prizes.remove(winner_prize)

print("=============LUCKY DRAW RESULT================")
print("winner:",winner)
print("prize",winner_prize)
print("\nbackup winner &  prizes")


for backup_winner,backup_prize in zip( backup_winners,backup_prizes):
  print(backup_winner,"---->",backup_prize)
print("\n====================================")
```
