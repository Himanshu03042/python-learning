```python
#DICTIONARY
student={
    "name":"himanshu",
    "age":"21",
    "course":"aiml"
}
print(student['name'])
print(student['age'])
student['city']="himachal"
print(student)
student["age"]="99"
print(student)
del student['age']
print(student)
student['college']="hgpi"
print(student)
# keys()   → keys
# values() → values
# items()  → keys + values
```

```python
#2nd PRACTICE
student={
    "name":"himanshu",
    "age":"35",
    "course":"cesss",
    "marks":99
}
print(student)
print(student.keys())
print(student.values())
print(student.items())
for key,values in student.items():
  print(key,values)
```

```python
#  get and pop
student={"name":"himanshu",
         "age":"20",
         "course":"ces"}

print(student.get('name'))
print(student.get('city',"not in dicet"))
print(student.pop('age'))
print(student)
```

```python
#NESTED DICt
students={"student1":{"name":"montiu",
                     "marks":87},
          "student2":{"name":"rohui",
                      "marks":88
                      }}
print(students)
print(students['student1'])
print(students['student2']['marks'])
print(students["student1"]["name"])
print(students["student1"]["marks"])
print(students["student2"]["name"])
print(students["student2"]["marks"])
```
