```python
class student:
  def __init__( self , name , age ):
    self.name=name
    self.age=age

student1=student("himanshu",21)
student2=student("rahul",34)
print(student1.name)
print(student1.age)
print(student2.name)
print(student2.age)
```

```python
class student:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def introduce(self):
    print("My name is:",self.name)
    print("My age is:",self.age)
  def study(self):
    print(self.name,"is studying")
student1=student("himanshu",25)
student1.introduce()
student1.study()
```

```python
# OOPS mini project :STUDENT MANAGEMENT
class student:
  def __init__(self,name,age,marks):
    self.name=name
    self.age=age
    self.marks=marks
  def show_result(self):
    print("my name is :-", self.name)
    print("my age is:-",self.age)
    print("my marks is:-",self.marks)
    if self.marks >=40:
      print("pass")
    else:
      print("fail")
  def show_grade(self):
    if self.marks >=90:
      print("A")
    elif self.marks >=75:
      print("B")
    elif self.marks >=60:
      print("C")
    else:
      print("fails")




student1=student("montu",25,90)
student1.show_result()
student2=student("rahul",55,39.9)
student2.show_result()
student3=student("naina",56,89)
student3.show_result()
student1.show_grade()
student2.show_grade()
student3.show_grade()
```
