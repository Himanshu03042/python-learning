```python
# CALCULATOR
def calculator(a,b,op):
  if op=='+':
    return a+b
  elif op=='-':
    return a-b
  elif op=='*':
    return a*b
  elif op=='/':
    return a/b
result = calculator(34,4,+)
print(result)
```

```python
# day 7 minin project
marks=[2,43,54,65,6,76,89,90,99,99,45]
def total_marks(marks):
  result=0
  for ma in marks:
    result=result+ma
  return result
print(total_marks(marks))

def avg_marks(marks):
  print("students",len(marks))
  return total_marks(marks)/len(marks)

print(avg_marks(marks))

def highest_marks(marks):
  result=marks[0]
  for ma in marks:
    if ma>result:
      result=ma
  return result
print(highest_marks(marks))

def lowest_marks(marks):
  result=marks[0]
  for ma in marks:
    if ma<result:
      result=ma
  return result
print(lowest_marks(marks))

def passed_marks(marks):
  result=[]
  for ma in marks:
    if ma>50:
      result.append(ma)
    else:
      pass
      print("below passing marks")
  return result
print(passed_marks(marks))



def grade_calculate(average):
  if average>=90:
    return("grade A")
  elif average>=75:
    return("grade B")
  elif average>=50:
    return("grade C")
  elif average<50:
    return("fail")
  else:
    pass
print(grade_calculate(5))

total =total_marks(marks)
avereage=avg_marks(marks)
highest=highest_marks(marks)
lowest=lowest_marks(marks)
passed=passed_marks(marks)
grade=grade_calculate(95)

print("total marks", total)
print("Total average",avereage)
print("highest marks",highest)
print("lowest marks",lowest)
print("passed marks",passed)
print(" actual grade",grade)
print("last line")
print(total,avereage,highest,lowest,passed,grade)
```
