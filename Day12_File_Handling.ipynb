```python
# file opening
with open("hello.txt","w") as file:
  file.write("hello I am learning python\n day12 File Handling")

with open("hello.txt","r") as file:
  data=file.read()
  print(data)
with open("hello.txt","a") as file:
  file.write("\n how are you")
with open("hello.txt", "r") as file:
    data = file.read()
    print(data)
```

```python
# Expense Tracker mini-project
print("\n ============Expense Tracker mini-project=============")
with open("expenses.txt","w") as file:
  file.write("food-- 50 \n travel:-- 58 \n stationary:-- 100")



while True:
    category=input("enter the category:- ")
    try:
      amount=int(input("enter the amount:-"))
    except ValueError:
      print("please enter the valid number:--")
      continue
    with open("expenses.txt","a")as file:
      file.write("\n" + category + " " + str(amount))
    choice=input("add other expense--(yes/no):-")
    while choice != "yes" and choice != "no":
      choice=input("Please enter yes/no:-")


    if choice =="no":
      break


total =0
with open("expenses.txt","r") as file:
  for lines in file:
    parts=lines.split()
    amount=parts[1]
    category=parts[0]
    amount=int(amount)
    total=total+amount
    print(category,":-",amount)
print("\n-------- EXPENSE SUMMARY --------")
print("total expenses:--",total)



```
