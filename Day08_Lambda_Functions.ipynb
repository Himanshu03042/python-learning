```python
# lmbda function
result =lambda x:x*2
print(result(3))
result=lambda x:x*x
print(result(3))
add=lambda a,b:a+b
print(add(4,6))
check=lambda x: "even" if x%2==0 else "odd"
print(check(7))
numbers=[2,4,6,8,1,0]
double=map(lambda x:x*2,numbers)
print(list(double))
f=filter(lambda x:x%2==0,numbers)
print(list(f))
```

```python
# number analyzer
numbers=[12,5,8,21,30,17,44]
a=map(lambda x:x*x,numbers)
b=filter(lambda x:  x%2==0 ,numbers)
c=map(lambda c:c*2,numbers)
print(list(a))
print(list(b))
print(list(c))
```
